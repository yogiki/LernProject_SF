анализ числовых признаков

```python
# Определение числовых признаков
numeric_columns = data.select_dtypes(include=['int64', 'float64']).columns.tolist()

print("Числовые признаки в датасете:")
print(", ".join(numeric_columns))

# Дополнительная информация о каждом числовом признаке
print("\nДетальная информация о числовых признаках:")
for col in numeric_columns:
    print(f"\n{col}:")
    print(f"  Тип: {data[col].dtype}")
    print(f"  Уникальных значений: {data[col].nunique()}")
    print(f"  Пропусков: {data[col].isnull().sum()} ({data[col].isnull().sum()/len(data)*100:.2f}%)")
    print(f"  Минимум: {data[col].min()}")
    print(f"  Максимум: {data[col].max()}")
    print(f"  Среднее: {data[col].mean():.2f}")
```

Проанализируйте датасет на наличие дублирующихся винных обзоров. Если дублирующиеся записи есть, удалите их.

В поле ниже введите, сколько дублирующихся записей вам удалось обнаружить.

```python


# Проверяем количество полных дубликатов (строк, которые полностью идентичны)
duplicate_count = data.duplicated().sum()

print(f"Количество полных дубликатов (идентичных строк): {duplicate_count}")

# Если есть дубликаты, удаляем их
if duplicate_count > 0:
    # Сохраняем количество строк до удаления
    before = len(data)
    
    # Удаляем дубликаты, оставляя первое вхождение
    data_cleaned = data.drop_duplicates()
    
    # Подсчитываем количество удаленных строк
    after = len(data_cleaned)
    removed = before - after
    
    print(f"Строк до удаления: {before}")
    print(f"Строк после удаления: {after}")
    print(f"Удалено дубликатов: {removed}")
    
    # Проверяем, что все дубликаты удалены
    remaining_duplicates = data_cleaned.duplicated().sum()
    print(f"Оставшихся дубликатов после очистки: {remaining_duplicates}")
    
    # При необходимости сохраняем очищенные данные
    # data_cleaned.to_csv('wine_cleaned.csv', index=False)
else:
    print("Дубликатов не обнаружено.")
```

Проверьте датасет на наличие пропусков в данных.

В каких из представленных признаках были обнаружены пропуски?

```python

# Список признаков для проверки
features_to_check = [
    'description',
    'designation', 
    'points',
    'price',
    'region_2',
    'taster_twitter_handle',
    'winery'
]

# Проверка пропусков в указанных признаках
missing_counts = {}

print("Проверка пропусков в данных:")
print("-" * 50)

for feature in features_to_check:
    missing = data[feature].isnull().sum()
    missing_counts[feature] = missing
    
    if missing > 0:
        print(f"{feature}: {missing} пропусков ({missing/len(data)*100:.2f}%)")
    else:
        print(f"{feature}: пропусков нет")

print("-" * 50)

# Определяем признаки с пропусками
features_with_missing = [feature for feature in features_to_check if missing_counts[feature] > 0]

if features_with_missing:
    print(f"Пропуски обнаружены в следующих признаках: {', '.join(features_with_missing)}")
else:
    print("Пропусков не обнаружено ни в одном из проверенных признаков")

# Дополнительно: общая статистика по пропускам
print(f"\nОбщее количество строк в датасете: {len(data)}")
```
